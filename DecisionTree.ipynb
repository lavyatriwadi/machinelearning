```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression , LogisticRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score , r2_score , mean_squared_error , confusion_matrix , classification_report
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import fetch_california_housing , load_iris , load_diabetes , load_breast_cancer
```

```python
data = load_iris(as_frame= True).frame
```

```python
data.head()
```

```python
x = data.iloc[:,[0,1,2,3]]
y = data.iloc[:,4]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt1 = DecisionTreeClassifier(criterion = 'gini' , random_state = 42)
dt2 = DecisionTreeClassifier(criterion = 'entropy' , random_state = 42)
```

```python
dt1.fit(x_train , y_train)
```

```python
dt2.fit(x_train , y_train)
```

```python
y_pred1 = dt1.predict(x_test)
y_pred2 = dt2.predict(x_test)
```

```python
print("Accuracyscore of gini" , accuracy_score(y_test , y_pred1))
print("Accuracyscore of entropt" , accuracy_score(y_test , y_pred2))
```

```python
confusion_matrix(y_test , y_pred1)
```

```python
confusion_matrix(y_test , y_pred2)
```

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(15,8))
plot_tree(dt1 ,
         feature_names = x.columns,
         class_names = ['setosa' , 'virginica' , 'versicolor'],
         filled = True)
plt.title("Decison tree based on gini")
plt.show()
```

```python
plt.figure(figsize=(15,8))
plot_tree(dt2 ,
         feature_names = x.columns,
         filled = True)
plt.title("Decison tree based on gini")
plt.show()
```
